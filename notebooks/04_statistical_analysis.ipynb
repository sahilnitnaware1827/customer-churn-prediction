# Customer Churn Prediction — Statistical Analysis

The EDA stage identified several patterns that may be associated with
customer churn.

In this stage, statistical methods are used to determine whether selected
relationships and differences are statistically significant.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats

data = pd.read_csv(
    "../data/cleaned/telco_customer_churn_cleaned.csv"
)

data.head()
```

```python
data["tenure"].median()
```

```python
data["tenure"].median()
```

```python
data["tenure"].std()
```

```python
data["tenure"].describe()
```

```python
tenure_no_churn = data.loc[
    data["Churn"] == "No",
    "tenure"
]
```

```python
tenure_churn = data.loc[
    data["Churn"] == "Yes",
    "tenure"
]
```

```python
t_stat, p_value = stats.ttest_ind(
    tenure_no_churn,
    tenure_churn,
    equal_var= False
)

print(f"\n T Stats : {t_stat}")
print(f"\n P value : {p_value}")
```

```python
contract_table = pd.crosstab(
    data["Contract"],
    data["Churn"]
)
contract_table
```

```python
chi2, p_value, dof, expected = stats.chi2_contingency(
    contract_table
)

print(f"\n Chi square contingency : {chi2}")
print(f"\n P Value : {p_value}")
print(f"\n Degree of Freedom : {dof}")
```

```python
internet_table = pd.crosstab(
    data["InternetService"],
    data["Churn"]
)

internet_table
```

```python
chi2, p_value, dof, expected = stats.chi2_contingency(
    internet_table
)

print("\n Chi-square statistic:", chi2)
print("\n P-value:", p_value)
```

## Statistical Analysis Summary

Statistical hypothesis tests were performed to validate relationships
identified during exploratory data analysis.

### Tests performed

1. Independent two-sample t-test
   - Used to test whether tenure differs significantly between churned
     and non-churned customers.

2. Chi-square test of independence
   - Used to test whether contract type is associated with customer churn.

3. Chi-square test of independence
   - Used to test whether internet service type is associated with churn.

A significance level of 0.05 was used for hypothesis testing.

The statistical results will be considered alongside the EDA findings
when selecting features for machine learning.
