# Customer Churn Prediction — Data Cleaning

The raw dataset was inspected during the data understanding stage.
This notebook focuses on identifying and resolving data-quality issues
before exploratory data analysis and model development.

```python
import pandas as pd 
import numpy as np 

data = pd.read_csv("../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
data.shape
```

```python
data.info()
```

```python
data["TotalCharges"] = data["TotalCharges"].str.strip()
data["TotalCharges"] = data["TotalCharges"].replace("", np.nan)
```

```python
data["TotalCharges"].isna().sum()
```

```python
data["TotalCharges"] = pd.to_numeric(
    data["TotalCharges"],
    errors= "coerce"
)
```

```python
data["TotalCharges"].dtypes
```

```python
data[data["TotalCharges"].isna()][
    ["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]
]
```

```python
data["TotalCharges"] = data["TotalCharges"].fillna(0)
```

```python
data["TotalCharges"].isna().sum()
```

```python
data["customerID"].duplicated().sum()
```

```python
categorical_columns = data.select_dtypes(include="str").columns

for column in categorical_columns:
    print(f"\n{column}")
    print(data[column].value_counts())
```
